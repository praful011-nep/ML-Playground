## Logistic Regression

Logistic Regression is a supervised machine learning algorithm used mainly for **classification**. It predicts the probability of an input belonging to a class, usually binary classes (0 or 1).

### 1. Sigmoid Function

The linear equation calculates a score:

$$
z = \theta_0 + \theta_1 x_1 + \cdots + \theta_n x_n
$$

The sigmoid function converts this score into a probability between 0 and 1:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

### 2. Classification

Usually, if \(p\geq0.5\), the model predicts class 1; otherwise, it predicts class 0. The threshold can be adjusted depending on the problem.

### 3. Log Loss Function

Log Loss measures the error in predicted probabilities:

$$
L = -\left[y\ln(\sigma(z)) + (1-y)\ln(1-\hat{y})\right]
$$

The model learns its coefficients by minimizing the average Log Loss. Lower Log Loss indicates better probability predictions, while confidently incorrect predictions receive a large penalty.

**Example:** Predicting whether a student will pass or fail an exam.
