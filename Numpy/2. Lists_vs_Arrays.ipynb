## Lists Vs Arrays

- Arrays are faster
- Arrays are convinient to work on
- Arrays occupy less memory

List and Array of same elements

```python
lista = range(100)
arr = np.arange(100)
```

### Memory Occupied

sys library used to calculate memory occupied by a single element of list

- Since 50 is one of the element of list, then multiplying the size of one element with lenght of list gives total memory occupied by the list

```python
print(sys.getsizeof(50)*len(lista))
```

Similarly of array

```python
print(arr.itemsize * arr.size)
```

### Speed Calculation

- time of before calculation is taken
- adding corresponding elements of both lists using zip function
- difference of time after the calculation and before calculation is taken i.e. time taken for calculation

```python
x = range(10000000)
y = range(10000000, 20000000)
```

```python
start_time = time.time()
c = [(x,y) for x,y in zip(x,y)]
lists= time.time()- start_time
print(lists)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)
```

```python
start_time = time.time()
c = a + b
array = time.time() - start_time
print(array)
```

```python
faster = ""
faster_by = 0.0
if array < lists:
    faster = "array"
    faster_by = ((lists - array)/lists)*100
elif lists < array:
    faster = "list"
    faster_by = ((array - lists)/array)*100
```
