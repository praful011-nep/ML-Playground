## Array Reshaping and Resizing

| Method | Method/ Attribute | Equivalent Function | Description |
|---|---|---|---|
| **Reshape** | `array.reshape(shape)` | `np.reshape(array, shape)` | Changes the shape of an array without changing its data. |
| **Resize** | `array.resize(shape)` | `np.resize(array, shape)` | Changes the size/shape of an array and stores the result in the same array. |
| **Ravel** | `array.ravel()` | `np.ravel(array)` | Converts a multidimensional array into a 1D array. |
| **Transpose** | `array.transpose()` / `array.T` | `np.transpose(array)` | Transposes an array by swapping its rows and columns (axes). |

```python
arr=np.arange(1,31)
arr
```

```python
arr = arr.reshape(5,6)
arr
```

```python
resize = np.resize(arr,(6,6))
resize
```

```python
arr.ravel()
```

```python
np.transpose(arr)
```

```python
arr.T
```

```python
arr.transpose()
```

## Stacking and Splitting Arrays

**Stacking Arrays**

- Combines two or more arrays into a single array.
- Arrays must have compatible dimensions.
- The stacked vectors are stacked in the form of matrix or vectors vertically (row wise) or horizontally (column wise)*  

| Function | Direction | Description |
|----------|-----------|-------------|
| `np.vstack((Matrix1, Matrix2))` | Vertical (Row-wise) | Stacks arrays vertically by adding rows. The resulting array has more rows while the number of columns remains the same. |
| `np.hstack((Matrix1, Matrix2))` | Horizontal (Column-wise) | Stacks arrays horizontally by adding columns. For 1D arrays, the result is a longer 1D array. |
| `np.column_stack((Matrix1, Matrix2))` | Horizontal (Column-wise) | Stacks 1D arrays as columns to create a 2D matrix. For 2D arrays, it behaves similarly to `np.hstack()`. |
| `np.stack((Matrix1, Matrix2), axis = )` | Specified by `axis` | Joins arrays along a new axis. The `axis` parameter determines the direction in which the arrays are stacked. |

- For 2D( and higher dimensional) arrays the result for column_stack and hstack are same

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
c = np.arange(6).reshape(2,3)
d = np.arange(10,16).reshape(2,3)
a
```

**Splitting Arrays**

- Splits an array into multiple sub-arrays.
- Returns a list of sub-arrays.
- Original array remains unchanged.

| Function | Direction | Description |
|----------|-----------|-------------|
| `np.split(array, sections, axis = ` | General | Splits an array into equal sub-arrays or at specified indices. |
| `np.vsplit(array, sections)` | Vertical (Row-wise) | Splits an array into multiple arrays along the rows. |
| `np.hsplit(array, sections)` | Horizontal (Column-wise) | Splits an array into multiple arrays along the columns. |

- `sections` - should be in such that the sub-arrays are evenly seperated
    - vsplit - any factor of row
    - hsplit - any factor of column
    - i.e. must be factor of dimension you're splitting

```python
ar = np.arange(0,20).reshape(10,2)
ar
```

- For a array of 10 x 2 dimension: 
    - vsplit - section - 1,2,5,10
    - hsplit - section - 1 ,2

```python
np.split(ar, 10, axis = 0)
```

```python
np.vsplit(ar,5) #Row-wise
```

```python
np.hsplit(ar, 2) #Column wise
```
