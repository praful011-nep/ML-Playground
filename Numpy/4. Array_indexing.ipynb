```python
import numpy as np
```

### Indexing and Slicing of Vectors

```python
arr=np.arange(11,21)
arr
```

### Indexing

- Indexing works just like lists (starts from 0)

```python
arr[2]
```

### Slicing

`Syntax` - *array*[*start : end* : *step*]  
- *start* - include, *end* - exclude
- *step* - increment value between elements
-  By default:
    - *start* : 0
    - *end* : end of list
    - *step* - 1
- whole array is returend if both start and end index are not given

```python
arr[:3]
```

```python
arr[:] # whole array is returned
```

```python
arr[1:5:2]
```

```python
arr[3::2]
```

### Indexing and Slicing of Matrix

```python
a=np.arange(1,31).reshape(6,5)
a
```

**Indexing**

- `Syntax` - *array*[*rows , columns*]

```python
a[5,2]
```

**Slicing**

`Syntax` - *array*[*row , column*]

- The sliced portion is returend in vector if only one dimension is given


```python
a
```

- Sliicing only rows
    - If only one index is given then that will target rows

```python
a[1]
```

- For a portion, slice rows as well as columns

```python
a[:2,2:4]
```

```python
a[4:,3:]
```

```python
a[3:,3:]
```

```python
a[:3:2,3:]
# 1st row and third row, 4th and 5th column
```

* Slicing a column:  
    - Returns anly column in vector form

```python
a[:,2]
```

### Advanced Indexing (Fancy Indexing)

- Numpy technique used to select multiple specific elements from an array by passing an array/list of indices

```python
n = np.arange(10,20)
n
```

```python
n[[1,5,8]] #Gives elements of desired index
```

```python
a
```

```python
a[[1,2,3]] #For 2D matrix
```

For element indexing in 2D matrix

```python
a[[1,3], [4,2]]
```

### Boolean Indexing

- Uses boolean mask i.e. True and False, to filter and extract specific elements(True) from a array
- Returns the boolean output of every index creating a new list.
- Only the True value can be extracted.

```python
ar=np.arange(11,21)
ar
```

```python
# List of boolean outputs
bool_index=ar%2==0
bool_index
```

```python
#Extracts only the True output indices.
ar=ar[bool_index]
ar
```

## Iterating over Array

- Iteratiing over vectors; works with general for loop
- Does not work with matrix, as for loop returns each row
- For iterating over n-d array, `np.iter` is used
- Nested loop is also a solution but not effecient

```python
arr
```

```python
for i in arr:
    print(i)
```

```python
a
```

```python
for i in a:
    for j in i:
        print(j)
```

This gives all the elements like vector

```python
for i in np.nditer(a):
    print(i)
```
