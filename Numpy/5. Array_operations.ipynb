# Array Operations

### Arithmetic Operations

- Performs element-wise operations on corresponding elements.
- Arrays must have the same shape or be broadcastable.
- There are equivalent fucntions used instead of symbols.

| Operation | Operator | Equivalent NumPy Function |
|-----------|:--------:|---------------------------|
| Addition | `+` | `np.add(a, b)` |
| Subtraction | `-` | `np.subtract(a, b)` |
| Multiplication | `*` | `np.multiply(a, b)` |
| Division | `/` | `np.divide(a, b)` |
| Floor Division | `//` | `np.floor_divide(a, b)` |
| Modulus / Remainder | `%` | `np.mod(a, b)` |
| Exponentiation | `**` | `np.power(a, b)` |

```python
a1=np.array([1,2,3,4,5])
a2=np.array([6,7,8,9,10])
```

Sum

```python
a1 + a2
```

```python
np.add(a1,a2)
```

Difference

```python
a1 - a2
```

Product

```python
np.multiply(a1,a2)
```

Division

```python
a1 / a2
```

```python
np.divide(a1,a2)
```

Floor Division

```python
a1 // a2
```

Power

```python
np.power(a1,a2)
```

### Broadcasting

Broadcasting is NumPy's ability to perform arithmetic operations on arrays of different shapes without explicitly copying data.  
It automatically expands the smaller array to match the shape of the larger array.
 
**Benefits**
- Eliminates the need for loops
- Faster and memory efficient
- Simplifies mathematical operations on arrays

**Broadcasting Rules**
1. If the dimensions are different, NumPy adds `1` to the left of the smaller array's shape.
2. Two dimensions are compatible if:
   - They are equal, **or**
   - One of them is `1`.
3. If the dimensions are not compatible, NumPy raises a **ValueError**.

```python
l=[1,2,3,4,5]
arr=np.array(l)
arr
```

```python
arr+10
```

```python
arr2=np.arange(1,16).reshape(5,3)
arr2
```

```python
arr2+10
```

```python
l=[10,20,30,40]
arr=np.array(l)
arr
```

```python
arr + 10
```

### Deep and Shallow Copy

**Shallow Copy**
- Shares the same memory as the original array.
- Changes made to one array are reflected in the other.

**Deep Copy**
- Creates a completely independent copy.
- Changes made to the copied array do not affect the original array.

```python
a=np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice *10
slice
```

```python
a
```

```python
a[4:6] *10 
```

```python
a
```

```python
b = a.copy() #Shallow Copy
# b=a     #Deep Copy
b[0] = 99 
```

```python
b
```

```python
a
```

### Matrix Operation 

| Operation | Operator | Equivalent NumPy Function/ Method |
|-----------|:--------:|---------------------------|
| Element-wise Multiplication (Broadcasting) | `*` | `np.multiply(a,b)` |
| Matrix Multiplication | `@` | `np.dot(a,b)` / `np.matmul(a,b)` |
| Transpose | `.T` | `np.transpose(a)` / `array.transpose()` |

```python
m=np.array([[1,2],[4,5]])
n=np.array([[7,8],[3,6]])
m
```

```python
n
```

```python
m * n #Broadcasting
```

```python
m @ n #Matrix Multiplication
```

```python
m.dot(n)
```
