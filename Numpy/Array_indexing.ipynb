# Indexing and Slicing of Vectors

```python
arr=np.arange(11,21)
arr
```

### Indexing

- Indexing works just like lists (starts from 0)

```python
arr[2]
```

### Slicing

`Syntax` - *array*[*start : end* ]  
- start - include, end - exclude
-  if starting value is not given the default is 0 and if end is not given array is sliced till last element
- whole array is returend if both start and end index are not given

```python
arr[:3]
```

```python
arr[:] # whole array is returned
```

**Steps is Slicing**

-  `Syntax` - *array*[ *start :: steps* ] 

```python
arr[3::2]
```

### Indexing and Slicing of Matrix

```python
a=np.arange(1,31).reshape(6,5)
a
```

**Indexing**

- `Syntax` - *array*[*rows , columns*]

```python
a[5,2]
```

**Slicing**

- The sliced portion is returend in vector if only one dimension is given


```python
a
```

- If only one index is given then that will target rows


```python
a[1]
```

- For a portion, slice rows as well as columns

```python
a[:2,2:4]
```

```python
a[3:,3:]
```

* Slicing a column:  
    * `Syntax` - *array*[*row , column*]

```python
# Returns anly column in vector form
a[:,2]
```

### Boolean Indexing

- Uses boolean mask i.e. True and False, to filter and extract specific elements(True) from a array
- Returns the boolean output of every index creating a new list.
- Only the True value can be extracted.

```python
ar=np.arange(11,21)
ar
```

```python
# List of boolean outputs
bool_index=ar%2==0
bool_index
```

```python
#Extracts only the True output indices.
ar=ar[bool_index]
ar
```
