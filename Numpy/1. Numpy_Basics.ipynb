# Numpy

NumPy (Numerical Python) is an open-source Python library used for numerical computing. It provides a powerful **N-dimensional array (ndarray)** object along with mathematical functions to perform fast and efficient operations on large datasets.

**Why Numpy over Python List**

- Faster than Python Lists
- Uses homogeneous data types
- Optimized for numerical computations
- Foundation for Pandas, Scikit-Learn, TensorFlow, etc.

`Vector` : One dimensional array  
`Matrix` : 2D array

### Importing Numpy

```python
import numpy as np
```

**2. Array Function**

**arange function**

- Creating a array within a range with evenly spaced values
    - `Syntax` - np.arange(*start, stop, step*)
- *stop* is exclusive
- By default: 
    - *start* - 0
    - *step* - 1
    - Only *stop* is mandatory    

```python
a=np.arange(2,22,2)
a
```

```python
a1 = np.arange(15)
a1
```

**Array of zeros and ones**

- `Syntax`
    - np.zeros((*rows,columns*))  
    - np.ones((*rows,columns*))

```python
b=np.zeros((3,3))
print(b)
o=np.ones((2,2))
o
```

**Array of Identity Matrix**

- A identity matrix is returned of desired order
    - `Syntax` : np.identity( *matrx_order* )

```python
i = np.identity(4)
i
```

**linspace function**

- Array of evenly spaced numbers over a specified interval
- np.linspace(*start, end, no_of_elements*)

```python
p=np.linspace(1,50,10)
p
```

**3. Random Array Generation (using Random Library)**

| Function | Distribution | Range / Mean | Description |
|----------|--------------|--------------|-------------|
| `np.random.rand(shape)` | Uniform Distribution | `0 ≤ x < 1` | Generates random floating-point numbers uniformly between 0 and 1. |
| `np.random.randn(shape)` | Standard Normal Distribution | Mean = `0`, Std = `1` (typically values between `-3` and `3`) | Generates random numbers from a standard normal (Gaussian) distribution. |
| `np.random.randint(low, high, size)` | Uniform Integer Distribution | `low ≤ x < high` | Generates random integers within the specified range. |

```python
np.random.rand(10)
```

```python
np.random.randn(10)
```

 Random Number within specific range
 - `Syntax` - .randint(*start, end, size*)

```python
np.random.randint(10,20,4) 
```

### Array Methods

| Method | Syntax | Description |
|--------|--------|-------------|
| Minimum | `array.min()` | Returns the smallest element. |
| Maximum | `array.max()` | Returns the largest element. |
| Sum | `array.sum()` or `np.sum(array)` | Returns the sum of all elements. |
| Sum (Axis-wise) | `np.sum(array, axis=0)` | Returns the column-wise sum. |
| Sum (Axis-wise) | `np.sum(array, axis=1)` | Returns the row-wise sum. |
| Mean | `array.mean()` | Returns the average of all elements. |
| Standard Deviation | `array.std()` | Returns the standard deviation. |
| Variance | `array.var()` | Returns the variance. |
| Index of Maximum | `array.argmax()` | Returns the index of the maximum element. |
| Index of Minimum | `array.argmin()` | Returns the index of the minimum element. |

```python
print(arr2.min())
print(arr2.max())
```

```python
arr2.sum()
```

```python
print(np.sum(arr2,axis=1)) #Row wise
print(np.sum(arr2,axis=0)) #Column wise
```

```python
print(arr2.mean())
print(arr2.std())
```

```python
print(arr2.argmax())
print(arr2.argmin())
```
