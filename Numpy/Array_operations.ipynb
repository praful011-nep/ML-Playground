### Broadcasting

Broadcasting is NumPy's ability to perform arithmetic operations on arrays of different shapes without explicitly copying data.  
It automatically expands the smaller array to match the shape of the larger array.
 
**Benefits**
- Eliminates the need for loops
- Faster and memory efficient
- Simplifies mathematical operations on arrays

**Broadcasting Rules**
1. If the dimensions are different, NumPy adds `1` to the left of the smaller array's shape.
2. Two dimensions are compatible if:
   - They are equal, **or**
   - One of them is `1`.
3. If the dimensions are not compatible, NumPy raises a **ValueError**.

```python
l=[1,2,3,4,5]
arr=np.array(l)
arr
```

```python
arr+10
```

```python
arr2=np.arange(1,16).reshape(5,3)
arr2
```

```python
arr2+10
```

```python
arr + 10
```

### Deep and Shallow Copy

**Shallow Copy**
- Shares the same memory as the original array.
- Changes made to one array are reflected in the other.

**Deep Copy**
- Creates a completely independent copy.
- Changes made to the copied array do not affect the original array.

```python
a=np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice *10
slice
```

```python
a
```

```python
a[4:6] *10 
```

```python
a
```

```python
b = a.copy() #Shallow Copy
# b=a     #Deep Copy
b[0] = 99 
```

```python
b
```

```python
a
```

### Advanced Array Manipulation

**Stacking Array**  
- Can't stack 2 matrix  but can stack or merge 2 vectors. 
- The stacked vectors are stacked in the form of matrix or vectors vertically (row wise) or horizontally (column wise)*  

| Function | Direction | Description |
|----------|-----------|-------------|
| `np.vstack((Matrix1, Matrix2))` | Vertical (Row-wise) | Stacks arrays vertically by adding rows. The resulting array has more rows while the number of columns remains the same. |
| `np.hstack((Matrix1, Matrix2))` | Horizontal (Column-wise) | Stacks arrays horizontally by adding columns. For 1D arrays, the result is a longer 1D array. |
| `np.column_stack((Matrix1, Matrix2))` | Horizontal (Column-wise) | Stacks 1D arrays as columns to create a 2D matrix. For 2D arrays, it behaves similarly to `np.hstack()`. |

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b)) #Vertically
```

```python
np.hstack((a,b)) #Horizontally
```

```python
np.column_stack((a,b))
```

**Splitting Arrays**

- *Matrix can be split in to number of parts*  


| Function | Direction | Description |
|----------|-----------|-------------|
| `np.split(array, sections)` | General | Splits an array into equal sub-arrays or at specified indices. |
| `np.vsplit(array, sections)` | Vertical (Row-wise) | Splits an array into multiple arrays along the rows. |
| `np.hsplit(array, sections)` | Horizontal (Column-wise) | Splits an array into multiple arrays along the columns. |

```python
c = np.arange(16).reshape(4, 4)
c
```

```python
np.split(c,4) #General
```

```python
np.hsplit(c,4)  #Col_wise
```

```python
np.vsplit(c,4) #Row_wise
```

```python
for i in c:
    print(i)  #As the array is already split 
```
