## Plotting Graphs in Numpy
- A library is used matplotlib(sub-library - pyplot), to generate graphs.
- `Syntax` : plt.plot(*array1, array2*)
- This generates a plot of array1 vs array2

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-40,40,100)
x
```

```python
y = np.sin(x)
y
```

```python
plt.plot(
    x,
    y
)
```
