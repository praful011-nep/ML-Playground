**Level 9**

- Create HasRedCard column (Yes/No).
- Create AgeGroup (<20, 20-25, 26-30, 30+).
- Create Efficiency = Goals / 90s.
- Create Cards = Yellow + Red.
- Create MinutesPerMatch = Min / MP.

```python
# Creating function to for extraction process.
def HasRedCard(rcrd):
    if rcrd > 0:
        result = 'Yes'
    else:
        result = 'No'
    return result     
```

```python
# Creating a new column with appliying the function above
df['HasRedCard']=df['CrdR'].apply(HasRedCard) 
df[df['HasRedCard']=='Yes'][['Player','Squad','CrdR','HasRedCard']]
```

```python
def age_grp(age):
    if age<20:
        group = '<20'
    elif age >= 20 and age <= 25:
        group = '20-25'
    elif age >=26 and age <= 30:
        group = '26-30'
    else:
        group = '30+'
    return group                
```

```python
df['AgeGroup'] = df['Age'].apply(age_grp)
df[['Player','AgeGroup']]
```

```python
qualified = df[df['Min'] == 90]
qualified = df[df['MP'] >= 5]
qualified["Effeciency"] = (qualified['Gls'] / qualified['Min'])*90
```

```python
df["Effeciency"] = qualified['Effeciency']
df['Effeciency']= df['Effeciency']
df[ df['Effeciency'] == df['Effeciency'].max()]
```

```python
df['Total_Cards'] = df['CrdR'] + df["CrdY"]
df[df['Total_Cards'] == df['Total_Cards'].max()]
```

```python
df['MinPerMatch'] = df['Min'] / df['MP']
```
