### Filtering Outlier

-  First calculate IQR  
    `Syntax`: np.percentile(*array*, *percentile_value*)

```python
Q1 = np.percentile(arr, 25)
Q2 = np.percentile(arr, 50)
Q3 = np.percentile(arr, 75)
```

```python
print(Q1)
print(Q2)
print(Q3)
```

```python
IQR = Q3 - Q1
IQR
```
