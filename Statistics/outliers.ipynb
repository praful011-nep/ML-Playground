```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

### Filtering Outlier

-  First calculate IQR  
    `Syntax`: np.percentile(*array*, *percentile_value*)

```python
Q1 = np.percentile(arr, 25)
Q2 = np.percentile(arr, 50)
Q3 = np.percentile(arr, 75)
```

```python
print(Q1)
print(Q2)
print(Q3)
```

```python
IQR = Q3 - Q1
print("IQR: ")
IQR
```

- Find Upper fence and Lower fence

```python
UF = Q3 + (1.5*IQR)
print("Upper fence: ")
UF
```

```python
LF = Q1 - (1.5*IQR)
print("Lower Fence: ")
LF
```

- Filtering data

```python
l = [] # where the filtered data is stored
for i in arr:
    if i >= LF and i <= UF:
        l.append(i)
```

```python
l = np.array(l)
```

```python
print("Array excluding Outlier: ")
l
```
