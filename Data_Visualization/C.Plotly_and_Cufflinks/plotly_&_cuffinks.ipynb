```python
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn as sns
import cufflinks as cf
from plotly.offline import iplot
import plotly.express as px
```

# Plotly

- **Definition:** Plotly is an interactive data visualization library used to create dynamic and web-based charts.
- Supports zooming, panning, hovering, and exporting plots.
- Built on top of JavaScript, making it ideal for dashboards and web applications.
- Plotly Express is the recommended high-level API for creating plots with minimal code.

# Plotly Express

- **Definition:** Plotly Express (`plotly.express`) is the high-level interface of Plotly used to create interactive visualizations with a single function call.
- Similar to Seaborn in terms of simple syntax.
- Automatically handles labels, legends, colors, and themes.
- Recommended over Cufflinks for new projects and learning Plotly.

# Why Not Cufflinks?

- Cufflinks was created to connect **Pandas** with **Plotly** before Plotly Express existed.
- Plotly Express now provides simpler syntax and more features.
- Cufflinks is no longer actively maintained and is rarely used in modern data science projects.
- Most tutorials, documentation, and industry projects use **Plotly Express** instead of Cufflinks.

**Loading dataset in Plotly Express**

px.data.*dataset_name*()

```python
tips = px.data.tips()
tips
```

# Plots

**Line Plot**

* Shows how a value changes over time

```python
fig = px.line(y=tips['total_bill']) 
fig.show()
```

**Scatter Plot**

* Shows relationship between 2 numeric valurables.

px.scatter( *data_frame* , x = , y = )
* dataframe should always be first.

```python
px.scatter(
    tips,
    x = 'total_bill',
    y = 'tip',
    
).show()
```

**Bar Plot**

* Compare value across different categories

```python
px.bar(
    tips,
    x = 'total_bill',
    y = 'day'
)
```

**Histogram**

* shows distribution of numerical variables

```python
px.histogram(
    tips,
    'total_bill'
    
)
```

**Box Plot**

* Summarizes distributions and higlight outliers

```python
px.box(
    tips,
    x='day',
    y='total_bill'
)
```

**Pie Chart**

* Show proportions or percentages of a total

px.pie( *data_frame*, name = *'col_name'*, value = "*data_based_on_name*")

```python
px.pie(
    tips,
    names='day'
)
```

```python
px.pie(
    tips,
    names='day',
    values= 'total_bill'
)
```

**Scatter Matrix**

* Lets you examine relationships among multiple numerical variables simultaneously.

px.scatter_matrix( *data_frame* )

```python
px.scatter_matrix(tips)
```

**Density Heatmap**

* Density heatmaps are ideal when scatter plots become crowded because they reveal where data points are concentrated rather than showing every individual point.

```python
px.density_heatmap(
    tips,
    x='total_bill',
    y='tip'
)
```

# Common Parameters

| Parameter | Description |
|-----------|-------------|
| `color` | Colors the data based on the values of a column. |
| `size` | Sets the marker size according to a numerical column. |
| `symbol` | Uses different marker shapes for different categories. |
| `hover_name` | Displays a highlighted label when hovering over a data point. |
| `hover_data` | Adds extra column values to the hover tooltip. |
| `facet_row` | Splits the plot into multiple rows based on a categorical column. |
| `facet_col` | Splits the plot into multiple columns based on a categorical column. |
| `title` | Adds a title to the figure. |
| `labels` | Renames axis titles, legends, or hover labels. |
| `opacity` | Controls the transparency of markers or bars. |
| `template` | Applies a predefined visual theme to the plot. |
| `width` | Sets the width of the figure in pixels. |
| `height` | Sets the height of the figure in pixels. |
| `nbins` | Specifies the number of bins in a histogram. |
| `marginal` | Adds a marginal plot (e.g., box, violin, or rug plot) alongside a histogram or scatter plot. |
| `names` | Specifies the categorical column used for slices in a pie chart. |
| `values` | Specifies the numerical column that determines the size of each pie slice. |
| `hole` | Creates a donut chart by setting the size of the hole in the center. |

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex",
    size="size",
    symbol="smoker",
    hover_name="day",
    hover_data=["time"],
    facet_col="time",
    title="Scatter Plot Example",
    labels={
        "total_bill": "Total Bill ($)",
        "tip": "Tip ($)"
    },
    opacity=0.7,
    template="plotly_white",
    width=900,
    height=500
)

fig.show()
```

```python
fig = px.histogram(
    tips,
    x="total_bill",
    color="sex",
    nbins=25,
    marginal="box",
    title="Histogram Example"
)

fig.show()
```

```python
fig = px.pie(
    tips,
    names="day",
    values="tip",
    color="day",
    title="Pie Chart Example",
    hole=0.4
)

fig.show()
```

# Updating Figure

* Plotly express plots can be updated without making another plot

| Function | Description |
|----------|-------------|
| `fig.update_layout()` | Updates the overall layout of the figure, such as the title, legend, template, size, and margins. |
| `fig.update_traces()` | Updates the properties of all traces, such as marker size, color, line style, and text. |
| `fig.update_xaxes()` | Customizes the x-axis by modifying its title, ticks, gridlines, range, and other axis properties. |
| `fig.update_yaxes()` | Customizes the y-axis by modifying its title, ticks, gridlines, range, and other axis properties. |

```python
fig = px.scatter(
    tips,
    x='total_bill',
    y='tip'
)
```

```python
fig.update_layout(
    title='Tips Dataset'
)
```

```python
fig.update_xaxes(title='Bill')

fig.update_yaxes(title='Tip')
```

```python
fig.update_traces(
    marker_size=12
)
```
