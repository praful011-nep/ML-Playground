# Regression Plots

- **Definition:** Regression plots visualize the relationship between two numerical variables by fitting a regression line to the data.
- Used to identify trends, correlations, and predict the dependent variable from the independent variable.
- Helps determine whether the relationship is positive, negative, or has no clear trend.
- Primarily used during Exploratory Data Analysis (EDA) before building regression models.
- The two main regression plots in Seaborn are:
  - `sns.regplot()` – Creates a regression plot for two numerical variables.
  - `sns.lmplot()` – Similar to `regplot()` but supports grouping with `hue` and creating multiple plots using `col` and `row`.

****lmplot****

* Creates a scatter plot with a fitted regression line.
* Supports grouping using `hue` and faceting using `row` and `col`.
* Best for comparing regression relationships across different categories.
* Returns a `FacetGrid` object. [`FacetGrid` : Grid of small subplots; splits data into different groups based on categories]
* `sns.lmplot(x = , y = , hue = , data = )`
* `Additional Parameters: `
    * markers = *marker_style* [List format for categorically divided data.]
    * scatter_kws = accepts a dictionary of keyword arguments that are passed directly to the underlying Matplotlib.




 Common scatter_kws keys:
| Key | Description |
|---|---|
| `"s"` | Controls marker size (numerical value). |
| `"alpha"` | Sets marker transparency (value between `0` and `1`). |
| `"color"` | Overrides the color of the scatter points. |
| `"edgecolor"` | Changes the outline color of the data points. |
| `"linewidths"` | Adjusts the thickness of the marker outlines. |

Analyze the relationship between total bill and tip using a regression plot.

```python
sns.lmplot(
    x = 'total_bill',
    y = 'tip',
    data = tips
)
```

FacetGrid Used.

Categorize the plot on the basis of 'smoker'

```python
sns.lmplot(
    x = 'total_bill',
    y = 'tip',
    data = tips,
    hue = 'smoker',
    palette = 'rainbow',
    markers = ['o','v'],
    scatter_kws = {'s': 100}
)
```

**Regression Plot /Regplot**

* Creates a scatter plot with a fitted regression line.
* Used to visualize the relationship between two numerical variables.
* Doesnot suppport hue, row or col.
* Fits best in subplots unlike lmplot.
* Returns a Matplotlib `Axes` object.
* `sns.regplot( x= , y = , data = )`

```python
sns.regplot(
    x = 'total_bill',
    y = 'tip',
    data = tips
)
```
