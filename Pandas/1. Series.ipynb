# Pandas

Pandas is an open-source Python library used for data manipulation, analysis, and processing. It provides powerful and flexible data structures to efficiently work with structured and tabular data.

### Characteristics of Pandas
- Built on top of NumPy.
- Fast and efficient for data analysis.
- Handles structured and tabular data.
- Provides powerful indexing and filtering.
- Supports data cleaning and preprocessing.
- Easily reads and writes files (CSV, Excel, JSON, SQL, etc.).
- Integrates well with Matplotlib, Seaborn, and Scikit-learn.

### Main Data Structure
- **Series** → One-dimensional labeled array.
- **DataFrame** → Two-dimensional labeled table (rows and columns).

### NumPy vs Pandas

| NumPy | Pandas |
|--------|---------|
| Works with arrays | Works with Series and DataFrames |
| Best for numerical computations | Best for data analysis and manipulation |
| Homogeneous data | Heterogeneous data |
| Faster for mathematical operations | Rich tools for handling real-world datasets |
| Foundation of Pandas | Built on top of NumPy |

# Pandas Series

A series is a  one-dimensional labeled array capable of holding any datatype. 

### Characteristics of Series

- One-dimensional labeled array.
- Can store any data type.
- Each element has an associated index (label).
- Default index starts from `0`.
- Mutable (values can be modified).
- Built on top of NumPy arrays.

```python
import numpy as np
import pandas as pd
```

### Creating Series

A Series can be created from:
- Python List
- NumPy Array
- Dictionary
- Scalar Value

`Syntax`: pd.Series(*data*, index = , dtype = , name = )

| Parameter | Description |
|-----------|-------------|
| `data` | List, array, dictionary, or scalar values |
| `index` | Labels for the elements |
| `dtype` | Specifies the data type |
| `name` | Assigns a name to the Series |

```python
labels=['a','b','c']
my_list=[10,20,30]
arr=np.array([10,20,30])
d={1:10,2:20,3:30}
```

```python
pd.Series(my_list, dtype = float)
```

`Scalar values` : Contains only one value


```python
pd.Series(100, index = labels)
```

For custom labels

```python
pd.Series(my_list,index=labels)
```

In dictionary,  
Keys becomes labels and Values become elements

```python
pd.Series(d, name = 'Using Dictionary')
```
