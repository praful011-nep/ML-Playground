# Data Frame

A DataFrame is a **two-dimensional labeled data structure** in Pandas that stores data in rows and columns.  
It can hold multiple data types in different columns, similar to a spreadsheet or SQL table.


### Characteristics
- Two-dimensional (rows and columns)
- Mutable (can add, remove, or modify data)
- Supports heterogeneous data types
- Each column is a Pandas Series
- Row labels are called **index**
- Column labels are called **columns**

### Series vs DataFrame

| Series | DataFrame |
|--------|-----------|
| One-dimensional | Two-dimensional |
| Single column | Multiple columns |
| One index | Rows and columns |
| Can store one data type | Different columns can have different data types |

```python
import pandas as pd
import numpy as np
```

### Creating a Dataframe

A DataFrame can be created from:

- Dictionary
- List of Lists
- NumPy Array
- List of Dictionaries
- CSV / Excel / SQL files

`Syntax` : pd.Dataframe( *data*, index = , columns = , dtype = )

| Parameter | Description |
|-----------|-------------|
| `data` | Dictionary, list, NumPy array, Series, or another DataFrame |
| `index` | Labels for the rows |
| `columns` | Labels for the columns |
| `dtype` | Specifies the data type for all columns (if possible) |

**Dictionary** 
- Creating dataframe from a dictionary, key becomes the header

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df=pd.DataFrame(data)
df
```

**List**  
- Creating dataframe from a list, header are default set as index  
- For header, columns = *list  having column title* is added in the function

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2=pd.DataFrame(data_list)
df2
```

```python
columns=["Name","Age","City","Salary"]
df2=pd.DataFrame(data_list,columns=columns, index=['First','Second','Third','Forth'])
df2
```

### Selecting and Indexing of columns

### Creating a new Columns

`Syntax` : *data_frame*["*Column_header* "] = [*list containing values*]

```python
df2["Designation"]=["Doctor","Eng","Doctor","Eng"]
df2
```

### Removing Rows or Columns

`Syntax` : *data_frame*.drop("*Column_name or Row_label*",axis=....)

drop() just makes a copy of the dataframe rather than doing operations on the actual dataframe  

Axis

- `axis=0` → Rows
- `axis=1` → Columns
 
inplace

- `False` (Default) → Returns a new DataFrame.
- `True` → Modifies the original DataFrame.

```python
df2.drop("Designation", axis=1)
```

```python
df2
```

```python
df2.drop("Designation", axis=1, inplace= True)
```

```python
df2
```

### Conditional Filtering

Conditional filtering is used to select rows from a DataFrame that satisfy a specified condition.


`Syntax` : *dataframe*[*dataframe['Column_Name'] condition*]

```python
df2
```

```python
#Display data of age > 30
df2[df2["Age"] > 30]
```

```python
df2[(df2["Age"] > 30) & (df2["City"]== "Paris")]
```
